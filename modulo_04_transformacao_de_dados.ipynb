```python
dados = [
    (1, "  João da Silva  ", "JOAO.SILVA@EMAIL.COM", "(71) 98888-7777", "123.456.789-00", "SSA-1"),
    (2, "maria   OLIVEIRA",  "maria@gmail.com",      "71 3333-4444",    "987.654.321-11", "REC-20"),
    (3, "  PEDRO santos",    None,                    "(81)99999-0000",  "111.222.333-44", "SSA-300"),
]
df_cli = spark.createDataFrame(dados, ["id", "nome", "email", "telefone", "cpf", "codigo"])
df_cli.show()
```

```python
# ---------- 1. REMOVER ESPAÇOS ----------
df_cli.select(
    "nome",
    F.trim("nome").alias("trim"),       # remove espaços dos dois lados
    F.ltrim("nome").alias("ltrim"),     # só do lado esquerdo
    F.rtrim("nome").alias("rtrim")      # só do lado direito
).show(truncate=False)
```

```python
# ---------- 2. CAIXA (MAIÚSCULA / minúscula) ----------
df_cli.select(
    "nome",
    F.upper("nome").alias("upper"),                  # TUDO MAIÚSCULO
    F.lower("nome").alias("lower"),                  # tudo minúsculo
    F.initcap(F.trim("nome")).alias("initcap")       # Primeira Letra De Cada Palavra Maiúscula
).show(truncate=False)
```

```python
# ---------- 3. TAMANHO ----------
df_cli.select(
    "nome",
    F.length("nome").alias("tamanho_com_espacos"),
    F.length(F.trim("nome")).alias("tamanho_limpo")
).show(truncate=False)
```

```python
# ---------- 4. CONCATENAR ----------
# concat: se QUALQUER coluna for nula, o resultado inteiro vira nulo.
# concat_ws: ignora os nulos e usa um separador entre os valores.
df_cli.select(
    "id", "email",
    F.concat(F.trim("nome"), F.lit(" ; "), "email").alias("concat"),
    F.concat_ws(" | ", F.trim("nome"), "email").alias("concat_ws")
).show(truncate=False)     # linha do Pedro: concat = null, concat_ws = só o nome
```

```python
# ---------- 5. PEGAR PEDAÇO DO TEXTO ----------
# substring(coluna, posição_inicial, quantidade). A posição começa em 1, não em 0.
df_cli.select(
    "codigo",
    F.substring("codigo", 1, 3).alias("sigla"),          # 3 primeiros caracteres
    F.substring("codigo", -2, 2).alias("ultimos_2")      # posição negativa conta do final
).show()
```

```python

# ---------- 6. QUEBRAR TEXTO (SPLIT) ----------
# split devolve um ARRAY. getItem(n) pega o elemento n (começa em 0).
df_cli.select(
    "codigo",
    F.split("codigo", "-").alias("partes"),
    F.split("codigo", "-").getItem(0).alias("sigla"),
    F.split("codigo", "-").getItem(1).alias("numero")
).show()
```

```python
df_cli.select(
    "cpf", "telefone", "codigo",
    F.replace("cpf", F.lit("."), F.lit("")).alias("cpf_sem_ponto"),            # remove o ponto (troca por vazio)
    F.replace("telefone", F.lit(" "), F.lit("")).alias("telefone_sem_espaco"),
    F.replace("codigo", F.lit("-"), F.lit("_")).alias("codigo_com_underline")
).show(truncate=False)
```

```python
# ARRAY = lista de valores numa célula | STRUCT = grupo de campos nomeados numa célula
dados = [
    (1, "João",  ["Python", "SQL", "Spark"], ("Salvador", "BA")),
    (2, "Maria", ["SQL"],                    ("Recife", "PE")),
    (3, "Pedro", [],                         ("Curitiba", "PR")),     # array vazio
    (4, "Ana",   None,                       None),                   # array e struct nulos
]
df = spark.createDataFrame(
    dados,
    "id INT, nome STRING, skills ARRAY<STRING>, endereco STRUCT<cidade: STRING, uf: STRING>"
)
df.printSchema()
df.show(truncate=False)
```

```python
# ---------- 1. EXPLODE ----------
# explode: uma linha por elemento do array. As outras colunas se repetem.
# ATENÇÃO: linhas com array vazio ou nulo SOMEM (Pedro e Ana desaparecem).
df.select("id", "nome", F.explode("skills").alias("skill")).show()
```

```python

# ---------- 2. EXPLODE_OUTER ----------
# explode_outer: mantém as linhas com array vazio ou nulo, gerando null no elemento.
df.select("id", "nome", F.explode_outer("skills").alias("skill")).show()
```

```python
# ---------- 3. POSEXPLODE ----------
# posexplode: além do elemento, devolve a posição dele no array (começa em 0).
df.select("id", "nome", F.posexplode("skills").alias("posicao", "skill")).show()
# existe também posexplode_outer
```

```python
# ---------- 4. SPLIT + EXPLODE ----------
# Caso comum: texto separado por vírgula. Primeiro split (vira array), depois explode.
df_txt = spark.createDataFrame([(1, "Python,SQL,Spark"), (2, "SQL")], ["id", "skills_txt"])
df_txt.show()
```

```python
df_txt.select("id", F.explode(F.split("skills_txt", ",")).alias("skill")).show()
```

```python
# ---------- 5. FUNÇÕES ÚTEIS DE ARRAY (sem explodir) ----------
df.select(
    "nome", "skills",
    F.size("skills").alias("qtd_skills"),                      # tamanho do array (para null, o resultado depende da configuração)
    F.array_contains("skills", "SQL").alias("sabe_sql"),       # o array contém o valo             # acessa por posição (começa em 0)
).show(truncate=False)
```
