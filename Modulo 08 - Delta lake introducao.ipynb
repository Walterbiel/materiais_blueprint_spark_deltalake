```python
from pyspark.sql import functions as F

df = spark.read.parquet(
    '/Volumes/blueprint_spark/bronze/bronze/vendas.parquet',
    mergeSchema=True,
)

caminho = "/Volumes/blueprint_spark/bronze/bronze/delta/vendas_delta"
```

```python
display(df)
```

```python
df.write.format("delta").mode("overwrite").save(caminho)
```

```python
df_delta = spark.read.format("delta").load(caminho)
display(df_delta)
print(df_delta.count())
```

```python
df.write.mode("overwrite").saveAsTable("blueprint_spark.bronze.vendas_delta")
```
